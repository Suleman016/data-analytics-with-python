# WHILE LOOP

```python
count = 1
while count <= 10 :
    print(count)
    count += 1
```

```python
count = 1
while count <= 10 :
    print("Hello",count)
    count += 1
```

```python
number = 8
guess = 0
while guess != number:
    guess = int(input("Enter a number:"))
    if guess == number:
        print("Your guess is right!")
        break

    elif guess > number:
        print("Your guess is too high!")
    else:
        print("Your guess is too low!")
```

```python
#sum of numbers from 1 to n 
n = int(input("Enter n:"))
sum = 0
i = 1
while i <= n:
    for i in range (n+1):
        sum = sum + i
        i += 1
    print("Sum of n numbers is:",sum)
```

```python
password = 12345
attempts = 1
while attempts <= 3 :
    user_password = int(input("Enter your password"))

    if user_password == password:
        print("Login successfull!")
        break

else:
    print(f"unsuccessfull, {3 - attempts}  attempts left")

    attempts += 1
```

```python
#factorial
n = int(input("Enter n:"))
ans = 1
count = 1
while count <= n:
    ans = ans * count 
    count += 1
    print(ans)
```
