```python
# Task -- calculate grade of student s

# 90+ A+
# 80 - 90 -- A
# 70 - 80 B 
# 60 - 70 - C

# define a function get_grade(marks)


students = [
    {"name": "Ali", "score": 92},
    {"name": "Sara", "score": 75},
    {"name": "Ahmed", "score": 55},
    {"name": "Zara", "score": 88}
]
```

```python
# Write get_grade(score) that returns a letter grade.
# Then loop over a list of students and print each result.

def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

students = [
    {"name": "Ali", "score": 92},
    {"name": "Sara", "score": 75},
    {"name": "Ahmed", "score": 55},
    {"name": "Zara", "score": 88}
]

for student in students:
    grade = get_grade(student["score"])
    print(f"{student['name']} got grade {grade}")
```

# Functions Scopes


```python
# scope means where a variable can be seen and used 

# local scope: variables created inside function - they only exists inside function, we can't call them outside 
```

```python
def sum(a,b):
    answer_1 =  a + b # variable initiated inside a function -- local scope
    return answer_1

sum(10,6)
print(answer_1)
```

```python
count = 10 

def increment():
    count = 100 # ik new variable banaya inside a function
    return count

increment()
print(count)
```

```python
count = 10 
# global scope

def increment():
    global count
    count = count + 10 # new variable initialize nhi kare ga 
    return count

increment()
print(count)
```
