1) A company decided to give bonus of 5% to employee if his/her year of service is more than 5 years.
Ask user for their salary and year of service and print the net bonus amount.

```python
salary = float(input("Enter your salary:"))
years = int(input("Enter years of service"))

if years > 5:
    bonus = salary * 5 / 100
else:
    bonus = 0

print("Bonus amount",bonus)

total_salary = salary + bonus
print("Total_salary",total_salary)
```

11) A shop will give discount of 10% if the cost of purchased quantity is more than 1000.
Ask user for quantity
Suppose, one unit will cost 100.
Judge and print total cost for user.

```python
quantity = int(input("Enter quantity: "))

price = quantity * 100

if price > 1000:
    discount = price * 10 / 100
    total_cost = price - discount
else:
    total_cost = price

print("Quantity:", quantity)
print("Total Cost:", total_cost)
```

12) A school has following rules for grading system:

a. Below 25 - F

b. 25 to 45 - E

c. 45 to 50 - D

d. 50 to 60 - C

e. 60 to 80 - B

f. Above 80 - A

Ask user to enter marks and print the corresponding grade.

```python
marks = float(input("Enter your marks: "))

if marks < 25:
    grade = "F"
elif marks < 45:
    grade = "E"
elif marks < 50:
    grade = "D"
elif marks < 60:
    grade = "C"
elif marks <= 80:
    grade = "B"
else:
    grade = "A"

print("Marks:", marks)
print("Grade:", grade)
```

14)A student will not be allowed to sit in exam if his/her attendence is less than 75%.

Take following input from user

- Number of classes held

- Number of classes attended.

And print

- percentage of class attended

- Is student is allowed to sit in exam or not.


```python
classes_held = int(input("Enter number of classes held: "))
print("Number of classes held is :",classes_held)
classes_attended = int(input("Enter number of classes attended: "))
print("Number of classes attended is :",classes_attended)

percentage = (classes_attended / classes_held) * 100

print("Attendance Percentage:", percentage, "%")

if percentage >= 75:
    print("Student is allowed to sit in the exam")
else:
    print("Student is not allowed to sit in the exam")
```

16) Write a program to check if a year is leap year or not.

If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
year = int(input("Enter a year: "))
print("Year is :",year)

if year % 400 == 0:
    print("Leap Year")
elif year % 100 == 0:
    print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not a Leap Year")
```

6) Write a program to calculate the electricity bill (accept number of unit from user) according to the following criteria :
     Unit                                                     Price  
uptp 100 units                                             no charge
Next 200 units                                              Rs 5 per unit
After 200 units                                             Rs 10 per unit
(For example if input unit is 350 than total bill amount is Rs.3500
(For example if input unit is 97 than total bill amount is Rs.0
(For example if input unit is 150 than total bill amount is Rs.750

```python
units = int(input("Enter electricity units: "))
print("Units are:",units)

if units <= 100:
    bill = 0

elif units <= 300:
    bill = (units - 100) * 5

else:
    bill = (200 * 5) + (units - 300) * 10

print("Electricity Bill: Rs.", bill)
```

13) Take input of age of 3 people by user and determine oldest and youngest among them.

```python
age1 = int(input("Enter age of first person: "))
print("Age of First person is:",age1)

age2 = int(input("Enter age of second person: "))
print("Age of second person is:",age2)

age3 = int(input("Enter age of third person: "))
print("Age of third person is:",age3)

oldest = max(age1, age2, age3)
youngest = min(age1, age2, age3)

print("Oldest age:", oldest)
print("Youngest age:", youngest)
```
