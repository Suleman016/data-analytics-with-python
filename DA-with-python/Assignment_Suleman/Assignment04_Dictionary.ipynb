Q4 — Count Frequency of Numbers
numbers = [1,2,2,3,3,3,4,4,4,4]
Create frequency dictionary.

```python
numbers = [1,2,2,3,3,3,4,4,4,4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] +1
    else:
        frequency[num] = 1
    print(frequency)
```

Q5 — Word Counter
Take sentence input and count word frequency.

```python
sentence = input("Enter a sentence")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

    print(frequency)
```

Q6 — Student Marks & Average
marks = {"Math":80, "English":70, "Physics":90}
Calculate total and average

```python
marks = {
    "Math": 80,
    "English": 70,
    "Physics": 90
}

total = 0

for mark in marks.values():
    total = total + mark

average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

Q7 — Highest Scoring Student
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
Find highest scoring student.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest = 0
name = ""

for student, marks in students.items():
    if marks > highest:
        highest = marks
        name = student

print("Highest scoring student is:", name)
print("Marks:", highest)
```
