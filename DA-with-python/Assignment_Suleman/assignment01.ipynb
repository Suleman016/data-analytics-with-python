3.Write a program in which use all the operators we can use in Python

```python
a = 10
b = 30
#arithmetic operations
print("Addition",a+b)
print("Subtraction",a-b)
print("Multiplication",a*b)
print("Division",a/b)
print("Modulus",a%b)


#comparison operators
print("\nEqual to",a==b)
print("not equal to",a!=b)
print("Greater than Equal to",a>=b)
print("less than Equal to",a<=b)
print("Greater than",a>b)
print("less than",a<b)

# Logical Operators
print("\nAND:", a > 5 and b < 5)
print("OR:", a > 15 or b < 5)
print("NOT:", not(a > 5))

# Bitwise Operators
print("\nBitwise AND:", a & b)
print("Bitwise OR:", a | b)
print("Bitwise XOR:", a ^ b)
print("Bitwise NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

```python
# Assignment Operators
x = 10
x += 2
print("x += 2:", x)
x = 10
x -= 2
print("x -= 2:", x)
x = 10
x *= 2
print("x *= 2:", x)
x = 10
x /= 2
print("x /= 2:", x)
x = 10
x //= 2
print("x //= 2:", x)
x = 10
x %= 3
print("x %= 3:", x)
x = 10
x **= 2
print("x **= 2:", x)
```

6.Take the radius of a circle from a variable and calculate its Area and Circumference (use pi = 3.14159)

```python
radius = 6
pi = 3.14159

area =pi*radius*radius
circumference = 2*pi*radius

print("Radius:",radius)
print("Area of the circle:", area)
print("Circumference of the circle:", circumference)
```
