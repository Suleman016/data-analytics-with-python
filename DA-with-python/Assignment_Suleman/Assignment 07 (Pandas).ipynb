1) Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
import pandas as pd

employees = {
    "Name": ["Suleman", "Shayan", "Kabir", "Hassan", "Rohan", "Anas"],
    "Age": [28, 35, 42, 31, 26, 38],
    "City": ["Karachi", "lahore", "Islamabad", "Quetta", "KPK", "Rawalpindi"],
    "Salary": [45000, 52000, 68000, 41000, 39000, 75000],
}
df = pd.DataFrame(employees)

print("First 3 rows:\n", df.head(3))
print("\nShape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
```

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
df["Bonus"] = df["Salary"] * 0.10
df["Total"] = df["Salary"] + df["Bonus"]
df = df.rename(columns={"City": "Location"})
df = df.drop(columns="Bonus")
print(df)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
missing_df = pd.DataFrame(
    {
        "Name": ["Sharukh", "Rafay", "Hanzala", "Yaseen"],
        "Age": [24, None, 31, 29],
        "Salary": [42000, 51000, None, 47000],
        "City": ["Karachi", "lahore", None, "Quetta"],
    }
)

print("Missing values per column:\n", missing_df.isna().sum())
numeric_columns = missing_df.select_dtypes(include="number").columns
missing_df[numeric_columns] = missing_df[numeric_columns].fillna(
    missing_df[numeric_columns].mean()
)
clean_df = missing_df.dropna()
print("\nAfter filling numeric NaNs with column means:\n", missing_df)
print("\nAfter dropping rows that still contain missing values:\n", clean_df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
sorted_df = df.sort_values(
    by=["Salary", "Age"], ascending=[False, True]
)
print("Sorted by Salary descending, then Age ascending:\n", sorted_df)
print("\nTop 3 highest-paid employees:\n", df.nlargest(3, "Salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
salary_by_city = df.groupby("Location").agg(
    Average_Salary=("Salary", "mean"),
    Employee_Count=("Salary", "count"),
    Maximum_Salary=("Salary", "max"),
)
print(salary_by_city)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
employees = pd.DataFrame(
    {
        "Employee": ["Shahrukh", "Rafay", "Hanzala", "Yaseen"],
        "DeptID": [1, 2, 2, 4],
    }
)
departments = pd.DataFrame(
    {
        "DeptID": [1, 2, 3],
        "Department": ["HR", "Engineering", "Sales"],
    }
)

inner_join = employees.merge(departments, on="DeptID", how="inner")
left_join = employees.merge(departments, on="DeptID", how="left")
print("Inner join:\n", inner_join)
print("\nLeft join:\n", left_join)

employees_extra = pd.DataFrame(
    {"Employee": ["Shahrukh", "Hanzala"], "DeptID": [1, 3]}
)
combined = pd.concat([employees, employees_extra], ignore_index=True)
combined = combined.drop_duplicates().reset_index(drop=True)
print("\nConcatenated employees with duplicates removed:\n", combined)
```
