1. Write a program that accepts a list from user and print the alternate element of list.

```python
numbers = []

n = int(input("Enter numbers "))

for i in range(n):
    value = int(input("Enter numbers "))
    numbers.append(value)

print("Alternate elements are ")

for i in range(0, len(numbers), 2):
    print(numbers[i])
```

2. Write a program that accepts a list from user. Your program should reverse the content of list and 
display it. Do not use reverse() method

```python
numbers = []

n = int(input("Enter numbers "))

for i in range(n):
    value = int(input("Enter numbers "))
    numbers.append(value)

print("Main list:", numbers)

reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

print("Reversed list:", reversed_list)
```

 3. Find and display the largest number of a list without using built-in function max(). Your program 
should ask the user to input values in list from keyboard.

```python
numbers = []

n = int(input("Enter number: "))

for i in range(n):
    value = int(input("Enter numbers: "))
    numbers.append(value)

largest = numbers[0]

for i in range(1, len(numbers)):
    if numbers[i] > largest:
        largest = numbers[i]

print("List:", numbers)
print("Largest number:", largest)
```

 4. Write a program that rotates the element of a list so that the element at the first index moves to the 
second index, the element in the second index moves to the third index, etc., and the element in the last 
index moves to the first index.

```python
numbers = []

n = int(input("Enter number: "))

for i in range(n):
    value = int(input("Enter number: "))
    numbers.append(value)

last = numbers[-1]

for i in range(len(numbers) - 1, 0, -1):
    numbers[i] = numbers[i - 1]

numbers[0] = last

print("Rotated list:", numbers)
```

 6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It 
should print the date in the form March 12, 2021.

```python
date = input("Enter date (mm/dd/yyyy): ")

month, day, year = date.split("/")

months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

print(months[int(month) -1], int(day), ",", year)
```

 7. Write a program with a function that accepts a string from keyboard and create a new string after 
converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." 
the output should be "Stop And Smell The Roses"

```python
def capitalize_words(text):
    return text.title()

sentence = input("Enter a sentence: ")

new_string = capitalize_words(sentence)

print("New String:", new_string)
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be 
like this : 
Sum of row 1 = 32 
Sum of row 2 = 31 
Sum of row 3 = 63

```python
matrix = [
    [2, 11, 7, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42]
]

for i in range(len(matrix)):
    total = 0

    for j in range(len(matrix[i])):
        total = total + matrix[i][j]

    print("Sum of row", i + 1, "=", total)
```

9. Write a program to add two matrices of size n x m.

```python
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]
matrix2 = [
    [7, 8, 9],
    [10, 11, 12]
]
result = []

for i in range(2):
    row = []

    for j in range(3):
        row.append(matrix1[i][j] + matrix2[i][j])

    result.append(row)

print("Result:")
for row in result:
    print(row)
```

10. Write a program to multiply two matrices

```python
matrix1 = [
    [1, 2],
    [3, 4]
]
matrix2 = [
    [5, 6],
    [7, 8]
]
result = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] = result[i][j] + matrix1[i][k] * matrix2[k][j]

print("Result:")

for row in result:
    print(row)
```
